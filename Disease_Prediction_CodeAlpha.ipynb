```python
# ---- DATASET 1: Breast Cancer (built into sklearn) ----
from sklearn.datasets import load_breast_cancer
import pandas as pd

cancer = load_breast_cancer()

# Convert to DataFrame
df_cancer = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df_cancer['target'] = cancer.target

print("Shape:", df_cancer.shape)
print("\nTarget values:", df_cancer['target'].value_counts().to_dict())
print("\nFirst 3 rows:")
df_cancer.head(3)
```

```python
# ---- DATASET 2: Diabetes (built into sklearn) ----
from sklearn.datasets import load_diabetes
# Note: sklearn's diabetes is regression, so we use the Pima dataset instead
# Let's download it directly

import urllib.request

url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
columns = ['Pregnancies','Glucose','BloodPressure','SkinThickness',
           'Insulin','BMI','DiabetesPedigree','Age','Outcome']

urllib.request.urlretrieve(url, "diabetes.csv")
df_diabetes = pd.read_csv("diabetes.csv", names=columns)

print("Shape:", df_diabetes.shape)
print("\nTarget values:", df_diabetes['Outcome'].value_counts().to_dict())
df_diabetes.head(3)
```

```python
# ---- DATASET 3: Heart Disease (UCI) ----
url = "https://raw.githubusercontent.com/sharmaroshan/Heart-UCI-Dataset/master/heart.csv"
df_heart = pd.read_csv(url)

print("Shape:", df_heart.shape)
print("\nTarget values:", df_heart['target'].value_counts().to_dict())
df_heart.head(3)
```

```python
import numpy as np

# These columns can't be 0 biologically
zero_not_allowed = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Replace 0s with NaN, then fill with median
for col in zero_not_allowed:
    df_diabetes[col] = df_diabetes[col].replace(0, np.nan)
    df_diabetes[col] = df_diabetes[col].fillna(df_diabetes[col].median())

print("Zeros fixed! Checking mins now:")
print(df_diabetes[zero_not_allowed].min())
```

```python
# Breast Cancer
X_cancer = df_cancer.drop('target', axis=1)
y_cancer = df_cancer['target']

# Diabetes
X_diabetes = df_diabetes.drop('Outcome', axis=1)
y_diabetes = df_diabetes['Outcome']

# Heart Disease
X_heart = df_heart.drop('target', axis=1)
y_heart = df_heart['target']

print("Features & targets separated ✅")
print(f"Cancer:   X={X_cancer.shape},   y={y_cancer.shape}")
print(f"Diabetes: X={X_diabetes.shape}, y={y_diabetes.shape}")
print(f"Heart:    X={X_heart.shape},    y={y_heart.shape}")
```

```python
from sklearn.model_selection import train_test_split

# Breast Cancer
X_cancer_train, X_cancer_test, y_cancer_train, y_cancer_test = train_test_split(
    X_cancer, y_cancer, test_size=0.2, random_state=42)

# Diabetes
X_diabetes_train, X_diabetes_test, y_diabetes_train, y_diabetes_test = train_test_split(
    X_diabetes, y_diabetes, test_size=0.2, random_state=42)

# Heart Disease
X_heart_train, X_heart_test, y_heart_train, y_heart_test = train_test_split(
    X_heart, y_heart, test_size=0.2, random_state=42)

print("Train/Test Split Done ✅")
print(f"Cancer   → Train: {X_cancer_train.shape}, Test: {X_cancer_test.shape}")
print(f"Diabetes → Train: {X_diabetes_train.shape}, Test: {X_diabetes_test.shape}")
print(f"Heart    → Train: {X_heart_train.shape}, Test: {X_heart_test.shape}")
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Cancer
X_cancer_train_sc = scaler.fit_transform(X_cancer_train)
X_cancer_test_sc  = scaler.transform(X_cancer_test)

# Diabetes
X_diabetes_train_sc = scaler.fit_transform(X_diabetes_train)
X_diabetes_test_sc  = scaler.transform(X_diabetes_test)

# Heart
X_heart_train_sc = scaler.fit_transform(X_heart_train)
X_heart_test_sc  = scaler.transform(X_heart_test)

print("Feature Scaling Done ✅")
print("Scaled data sample (Cancer, first row):")
print(X_cancer_train_sc[0])
```

```python
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score

print("All models imported ✅")
```

```python
# Dictionary to store all results
results = {}

# ---- BREAST CANCER ----
models_cancer = {
    'SVM':               SVC(kernel='rbf', random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=10000, random_state=42),
    'Random Forest':     RandomForestClassifier(n_estimators=100, random_state=42),
    'XGBoost':           XGBClassifier(eval_metric='logloss', random_state=42)
}

results['Breast Cancer'] = {}
for name, model in models_cancer.items():
    model.fit(X_cancer_train_sc, y_cancer_train)
    preds = model.predict(X_cancer_test_sc)
    acc = accuracy_score(y_cancer_test, preds)
    results['Breast Cancer'][name] = round(acc * 100, 2)
    print(f"{name:25s} → Accuracy: {acc*100:.2f}%")
```

```python
# ---- DIABETES ----
models_diabetes = {
    'SVM':               SVC(kernel='rbf', random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=10000, random_state=42),
    'Random Forest':     RandomForestClassifier(n_estimators=100, random_state=42),
    'XGBoost':           XGBClassifier(eval_metric='logloss', random_state=42)
}

results['Diabetes'] = {}
for name, model in models_diabetes.items():
    model.fit(X_diabetes_train_sc, y_diabetes_train)
    preds = model.predict(X_diabetes_test_sc)
    acc = accuracy_score(y_diabetes_test, preds)
    results['Diabetes'][name] = round(acc * 100, 2)
    print(f"{name:25s} → Accuracy: {acc*100:.2f}%")
```

```python
# ---- HEART DISEASE ----
models_heart = {
    'SVM':               SVC(kernel='rbf', random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=10000, random_state=42),
    'Random Forest':     RandomForestClassifier(n_estimators=100, random_state=42),
    'XGBoost':           XGBClassifier(eval_metric='logloss', random_state=42)
}

results['Heart Disease'] = {}
for name, model in models_heart.items():
    model.fit(X_heart_train_sc, y_heart_train)
    preds = model.predict(X_heart_test_sc)
    acc = accuracy_score(y_heart_test, preds)
    results['Heart Disease'][name] = round(acc * 100, 2)
    print(f"{name:25s} → Accuracy: {acc*100:.2f}%")
```

```python
summary_df = pd.DataFrame(results)
print("\n===== ACCURACY SUMMARY (%) =====")
print(summary_df)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score

# Store predictions for all datasets
all_preds = {}

# Breast Cancer - retrain to get predictions
final_models_cancer = {
    'SVM':                SVC(kernel='rbf', random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=10000, random_state=42),
    'Random Forest':      RandomForestClassifier(n_estimators=100, random_state=42),
    'XGBoost':            XGBClassifier(eval_metric='logloss', random_state=42)
}

all_preds['Breast Cancer'] = {}
for name, model in final_models_cancer.items():
    model.fit(X_cancer_train_sc, y_cancer_train)
    all_preds['Breast Cancer'][name] = model.predict(X_cancer_test_sc)

# Diabetes
final_models_diabetes = {
    'SVM':                SVC(kernel='rbf', random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=10000, random_state=42),
    'Random Forest':      RandomForestClassifier(n_estimators=100, random_state=42),
    'XGBoost':            XGBClassifier(eval_metric='logloss', random_state=42)
}

all_preds['Diabetes'] = {}
for name, model in final_models_diabetes.items():
    model.fit(X_diabetes_train_sc, y_diabetes_train)
    all_preds['Diabetes'][name] = model.predict(X_diabetes_test_sc)

# Heart Disease
final_models_heart = {
    'SVM':                SVC(kernel='rbf', random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=10000, random_state=42),
    'Random Forest':      RandomForestClassifier(n_estimators=100, random_state=42),
    'XGBoost':            XGBClassifier(eval_metric='logloss', random_state=42)
}

all_preds['Heart Disease'] = {}
for name, model in final_models_heart.items():
    model.fit(X_heart_train_sc, y_heart_train)
    all_preds['Heart Disease'][name] = model.predict(X_heart_test_sc)

print("All predictions stored ✅")
```

```python
fig, axes = plt.subplots(3, 4, figsize=(20, 15))

datasets = {
    'Breast Cancer': y_cancer_test,
    'Diabetes':      y_diabetes_test,
    'Heart Disease': y_heart_test
}

model_names = ['SVM', 'Logistic Regression', 'Random Forest', 'XGBoost']

for row, (dataset, y_test) in enumerate(datasets.items()):
    for col, model_name in enumerate(model_names):
        cm = confusion_matrix(y_test, all_preds[dataset][model_name])
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    ax=axes[row][col], cbar=False)
        axes[row][col].set_title(f'{dataset}\n{model_name}', fontsize=9)
        axes[row][col].set_xlabel('Predicted')
        axes[row][col].set_ylabel('Actual')

plt.suptitle('Confusion Matrices — All Models & Datasets',
             fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

```python
print("=" * 55)
print("BREAST CANCER — Best Model: SVM")
print("=" * 55)
print(classification_report(y_cancer_test,
      all_preds['Breast Cancer']['SVM'],
      target_names=['Malignant', 'Benign']))

print("=" * 55)
print("DIABETES — Best Model: Logistic Regression")
print("=" * 55)
print(classification_report(y_diabetes_test,
      all_preds['Diabetes']['Logistic Regression'],
      target_names=['No Diabetes', 'Diabetes']))

print("=" * 55)
print("HEART DISEASE — Best Model: SVM")
print("=" * 55)
print(classification_report(y_heart_test,
      all_preds['Heart Disease']['SVM'],
      target_names=['No Disease', 'Disease']))
```
